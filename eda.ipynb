```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(data['SalePrice'], bins=50, kde=True, ax=axes[0])
axes[0].set_title('Распределение SalePrice')
sns.boxplot(y=data['SalePrice'], ax=axes[1])
axes[1].set_title('Ящик с усами для SalePrice')
plt.show()
```

```python
data['SalePrice_log'] = np.log1p(data['SalePrice'])
sns.histplot(data['SalePrice_log'], bins=50, kde=True)
plt.title('Распределение логарифма SalePrice')
plt.show()
```

```python
plt.figure(figsize=(22, 8))
sns.heatmap(data.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Карта пропусков')
plt.show()
```

```python
numeric_cols = data.select_dtypes(include=[np.number]).columns
corr = data[numeric_cols].corr()

corr_with_price = corr['SalePrice_log'].sort_values(ascending=False)
print(corr_with_price)

top_features = corr_with_price.head(11).index
plt.figure(figsize=(10, 8))
sns.heatmap(data[top_features].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Корреляция топ-10 признаков с ценой')
plt.show()
```

```python
cat_cols = data.select_dtypes(include=['object']).columns

for col in cat_cols:
    if data[col].nunique() <= 15:
        plt.figure(figsize=(10, 4))
        sns.boxplot(x=col, y='SalePrice', data=data)
        plt.xticks(rotation=45)
        plt.title(f'Зависимость SalePrice от {col}')
        plt.show()
    else:
        avg_price = data.groupby(col)['SalePrice'].mean().sort_values()
        plt.figure(figsize=(10, 6))
        avg_price.plot(kind='bar')
        plt.title(f'Средняя цена по {col}')
        plt.xticks(rotation=45)
        plt.show()
```
