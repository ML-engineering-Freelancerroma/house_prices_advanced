```python
import pandas as pd


data = pd.read_csv('train.csv')

data.drop('Id', axis=1, inplace=True)

df_clean = data.copy()

# Заполнение пропусков
df_clean['Alley'] = df_clean['Alley'].fillna('NoAlley')
df_clean['PoolQC'] = df_clean['PoolQC'].fillna('NoPool')
df_clean['Fence'] = df_clean['Fence'].fillna('NoFence')
df_clean['MiscFeature'] = df_clean['MiscFeature'].fillna('None')

df_clean['MasVnrType'] = df_clean['MasVnrType'].fillna('None')
df_clean['MasVnrArea'] = df_clean['MasVnrArea'].fillna(0)
```

```python
bsmt_cols = ['BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2']
for col in bsmt_cols:
    df_clean[col] = df_clean[col].fillna('NoBasement')
```

```python
bsmt_num_cols = ['BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF']
for col in bsmt_num_cols:
    df_clean[col] = df_clean[col].fillna(0)
```

```python
df_clean['FireplaceQu'] = df_clean['FireplaceQu'].fillna('NoFireplace')
```

```python
garage_cols = ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond']
for col in garage_cols:
    df_clean[col] = df_clean[col].fillna('NoGarage')
```

```python
df_clean['GarageYrBlt'] = df_clean['GarageYrBlt'].fillna(0)
```

```python
df_clean['GarageCars'] = df_clean['GarageCars'].fillna(0)
df_clean['GarageArea'] = df_clean['GarageArea'].fillna(0)
```

```python
df_clean['LotFrontage'] = df_clean.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))
```

```python
df_clean['Electrical'] = df_clean['Electrical'].fillna(df_clean['Electrical'].mode()[0])
```

```python
df_clean['Utilities'] = df_clean['Utilities'].fillna(df_clean['Utilities'].mode()[0])
```

```python
df_clean['Functional'] = df_clean['Functional'].fillna('Typ')
```

```python
df_clean['SaleType'] = df_clean['SaleType'].fillna(df_clean['SaleType'].mode()[0])
```

```python
quality_mapping = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'NoBasement': 0, 'NoFireplace': 0, 'NoGarage': 0, 'NoPool': 0, 'NoFence': 0}

exposure_mapping = {'Gd': 4, 'Av': 3, 'Mn': 2, 'No': 1, 'NoBasement': 0}

fin_mapping = {'GLQ': 6, 'ALQ': 5, 'BLQ': 4, 'Rec': 3, 'LwQ': 2, 'Unf': 1, 'NoBasement': 0}

garage_finish_mapping = {'Fin': 3, 'RFn': 2, 'Unf': 1, 'NoGarage': 0}

functional_mapping = {'Typ': 7, 'Min1': 6, 'Min2': 5, 'Mod': 4, 'Maj1': 3, 'Maj2': 2, 'Sev': 1, 'Sal': 0}

fence_mapping = {'GdPrv': 4, 'MnPrv': 3, 'GdWo': 2, 'MnWw': 1, 'NoFence': 0}

lotshape_mapping = {'Reg': 4, 'IR1': 3, 'IR2': 2, 'IR3': 1}

landcontour_mapping = {'Lvl': 4, 'Bnk': 3, 'HLS': 2, 'Low': 1}

utilities_mapping = {'AllPub': 4, 'NoSewr': 3, 'NoSeWa': 2, 'ELO': 1}

landslope_mapping = {'Gtl': 3, 'Mod': 2, 'Sev': 1}

paveddrive_mapping = {'Y': 3, 'P': 2, 'N': 1}

centralair_mapping = {'Y': 2, 'N': 1}
```
