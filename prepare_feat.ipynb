```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(data['SalePrice'], bins=50, kde=True, ax=axes[0])
axes[0].set_title('Распределение SalePrice')
sns.boxplot(y=data['SalePrice'], ax=axes[1])
axes[1].set_title('Ящик с усами для SalePrice')
plt.show()
```

```python
data['SalePrice_log'] = np.log1p(data['SalePrice'])
sns.histplot(data['SalePrice_log'], bins=50, kde=True)
plt.title('Распределение логарифма SalePrice')
plt.show()
```

```python
plt.figure(figsize=(22, 8))
sns.heatmap(data.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Карта пропусков')
plt.show()
```
