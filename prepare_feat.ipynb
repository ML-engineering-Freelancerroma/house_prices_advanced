```python
import pandas as pd
import numpy as np


data = pd.read_csv('train.csv')

data.drop('Id', axis=1, inplace=True)

df_clean = data.copy()

# Заполнение пропусков
df_clean['Alley'] = df_clean['Alley'].fillna('NoAlley')
df_clean['PoolQC'] = df_clean['PoolQC'].fillna('NoPool')
df_clean['Fence'] = df_clean['Fence'].fillna('NoFence')
df_clean['MiscFeature'] = df_clean['MiscFeature'].fillna('None')
df_clean['MasVnrType'] = df_clean['MasVnrType'].fillna('None')
df_clean['MasVnrArea'] = df_clean['MasVnrArea'].fillna(0)

bsmt_cols = ['BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2']
for col in bsmt_cols:
    df_clean[col] = df_clean[col].fillna('NoBasement')

bsmt_num_cols = ['BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF']
for col in bsmt_num_cols:
    df_clean[col] = df_clean[col].fillna(0)

df_clean['FireplaceQu'] = df_clean['FireplaceQu'].fillna('NoFireplace')

garage_cols = ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond']
for col in garage_cols:
    df_clean[col] = df_clean[col].fillna('NoGarage')

df_clean['GarageYrBlt'] = df_clean['GarageYrBlt'].fillna(0)

df_clean['GarageCars'] = df_clean['GarageCars'].fillna(0)
df_clean['GarageArea'] = df_clean['GarageArea'].fillna(0)

df_clean['LotFrontage'] = df_clean.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

df_clean['Electrical'] = df_clean['Electrical'].fillna(df_clean['Electrical'].mode()[0])

df_clean['Utilities'] = df_clean['Utilities'].fillna(df_clean['Utilities'].mode()[0])

df_clean['Functional'] = df_clean['Functional'].fillna('Typ')

df_clean['SaleType'] = df_clean['SaleType'].fillna(df_clean['SaleType'].mode()[0])
```

```python
print(df_clean.isnull().sum().sum())
```

```python
quality_mapping = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'NoBasement': 0, 'NoFireplace': 0, 'NoGarage': 0, 'NoPool': 0, 'NoFence': 0}
exposure_mapping = {'Gd': 4, 'Av': 3, 'Mn': 2, 'No': 1, 'NoBasement': 0}
fin_mapping = {'GLQ': 6, 'ALQ': 5, 'BLQ': 4, 'Rec': 3, 'LwQ': 2, 'Unf': 1, 'NoBasement': 0}
garage_finish_mapping = {'Fin': 3, 'RFn': 2, 'Unf': 1, 'NoGarage': 0}
functional_mapping = {'Typ': 7, 'Min1': 6, 'Min2': 5, 'Mod': 4, 'Maj1': 3, 'Maj2': 2, 'Sev': 1, 'Sal': 0}
fence_mapping = {'GdPrv': 4, 'MnPrv': 3, 'GdWo': 2, 'MnWw': 1, 'NoFence': 0}
lotshape_mapping = {'Reg': 4, 'IR1': 3, 'IR2': 2, 'IR3': 1}
landcontour_mapping = {'Lvl': 4, 'Bnk': 3, 'HLS': 2, 'Low': 1}
utilities_mapping = {'AllPub': 4, 'NoSewr': 3, 'NoSeWa': 2, 'ELO': 1}
landslope_mapping = {'Gtl': 3, 'Mod': 2, 'Sev': 1}
paveddrive_mapping = {'Y': 3, 'P': 2, 'N': 1}
centralair_mapping = {'Y': 2, 'N': 1}
```

```python
df_clean['ExterQual'] = df_clean['ExterQual'].map(quality_mapping)
df_clean['ExterCond'] = df_clean['ExterCond'].map(quality_mapping)
df_clean['BsmtQual'] = df_clean['BsmtQual'].map(quality_mapping)
df_clean['BsmtCond'] = df_clean['BsmtCond'].map(quality_mapping)
df_clean['HeatingQC'] = df_clean['HeatingQC'].map(quality_mapping)
df_clean['KitchenQual'] = df_clean['KitchenQual'].map(quality_mapping)
df_clean['FireplaceQu'] = df_clean['FireplaceQu'].map(quality_mapping)
df_clean['GarageQual'] = df_clean['GarageQual'].map(quality_mapping)
df_clean['GarageCond'] = df_clean['GarageCond'].map(quality_mapping)
df_clean['PoolQC'] = df_clean['PoolQC'].map(quality_mapping)
df_clean['BsmtExposure'] = df_clean['BsmtExposure'].map(exposure_mapping)
df_clean['BsmtFinType1'] = df_clean['BsmtFinType1'].map(fin_mapping)
df_clean['BsmtFinType2'] = df_clean['BsmtFinType2'].map(fin_mapping)
df_clean['GarageFinish'] = df_clean['GarageFinish'].map(garage_finish_mapping)
df_clean['Functional'] = df_clean['Functional'].map(functional_mapping)
df_clean['Fence'] = df_clean['Fence'].map(fence_mapping)
df_clean['LotShape'] = df_clean['LotShape'].map(lotshape_mapping)
df_clean['LandContour'] = df_clean['LandContour'].map(landcontour_mapping)
df_clean['Utilities'] = df_clean['Utilities'].map(utilities_mapping)
df_clean['LandSlope'] = df_clean['LandSlope'].map(landslope_mapping)
df_clean['PavedDrive'] = df_clean['PavedDrive'].map(paveddrive_mapping)
df_clean['CentralAir'] = df_clean['CentralAir'].map(centralair_mapping)
```

```python
df_clean['TotalSF'] = df_clean['TotalBsmtSF'] + df_clean['1stFlrSF'] + df_clean['2ndFlrSF'] + df_clean['LowQualFinSF']
df_clean['TotalBath'] = df_clean['FullBath'] + 0.5*df_clean['HalfBath'] + df_clean['BsmtFullBath'] + 0.5*df_clean['BsmtHalfBath']
df_clean['Age'] = df_clean['YrSold'] - df_clean['YearBuilt']
df_clean['AgeRemod'] = df_clean['YrSold'] - df_clean['YearRemodAdd']

df_clean['HasBasement'] = (df_clean['TotalBsmtSF'] > 0).astype(int)
df_clean['HasGarage'] = (df_clean['GarageArea'] > 0).astype(int)
df_clean['HasFireplace'] = (df_clean['Fireplaces'] > 0).astype(int)
df_clean['HasPool'] = (df_clean['PoolArea'] > 0).astype(int)
df_clean['GarageYrBlt'] = df_clean['GarageYrBlt'].astype(int)

df_clean['GarageAge'] = np.where(
    df_clean['HasGarage'] == 1,
    df_clean['YrSold'] - df_clean['GarageYrBlt'],
    0
)
```

```python
df_clean['Qual*GrLiv'] = df_clean['OverallQual'] * df_clean['GrLivArea']
```

```python
cat_cols = ['MSSubClass', 'MoSold', 'YrSold', 'Alley', 'MiscFeature',
            'MSZoning', 'Street', 'LotConfig', 'Neighborhood', 'Condition1', 'Condition2',
            'BldgType', 'HouseStyle', 'RoofStyle', 'RoofMatl', 'Exterior1st', 'Exterior2nd',
            'MasVnrType', 'Foundation', 'Heating', 'Electrical', 'GarageType',
            'SaleType', 'SaleCondition']

for col in cat_cols:
    df_clean[col] = df_clean[col].astype(str)
```

```python
df_clean['MasVnrArea'] = df_clean['MasVnrArea'].astype(int)
df_clean['GarageAge']  = df_clean['GarageAge'].astype(int)
```
